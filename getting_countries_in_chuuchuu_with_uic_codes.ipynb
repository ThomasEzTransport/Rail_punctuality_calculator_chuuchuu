Including countries of each stop in Chuuchuu data

```python
import pandas as pd
```

```python
uic_codes = pd.read_csv("UIC_country_codes.csv", sep =";")
chuuchuu = pd.read_csv("delay_records_2026-02-26_to_2026-03-04.csv")
```

```python
chuuchuu.head()
```

```python
chuuchuu["uicCodeStop"] = chuuchuu["deutscheBahnStopId"].astype(str).str[:2]
```

```python
chuuchuu["uicCodeStop"].head(5)
```

```python
chuuchuu["uicCodeStop"] = chuuchuu["uicCodeStop"].astype(str)
uic_codes["Numerical code"] = uic_codes["Numerical code"].astype(str)
```

```python
chuuchuu = chuuchuu.merge(
    uic_codes[["Numerical code", "Alphabetical code"]],
    how="left",                       # keep all rows in chuuchuu
    left_on="uicCodeStop",            # column in chuuchuu
    right_on="Numerical code"         # column in uic_codes
)
chuuchuu = chuuchuu.drop(columns=["Numerical code"])
```

```python
chuuchuu = chuuchuu.rename(columns={"Alphabetical code": "countryStop"})
```
